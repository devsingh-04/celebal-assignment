```python
import importlib.util

def import_from_file(module_name, file_path):
    spec = importlib.util.spec_from_file_location(module_name, file_path)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module

# 📍 Give full path to your script
module_path = os.path.abspath(os.path.join(os.getcwd(), '..', 'scripts', 'model_training.py'))
model_training = import_from_file("model_training", module_path)

# ✅ Now call your function
train_and_evaluate = model_training.train_and_evaluate
```

```python
# ✅ Manual import from scripts folder
import importlib.util
import os

def import_custom_module(name):
    path = os.path.abspath(os.path.join(os.getcwd(), '..', 'scripts', f'{name}.py'))
    spec = importlib.util.spec_from_file_location(name, path)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module

# Load all script modules manually
model_training = import_custom_module("model_training")
hyperparameter_tuning = import_custom_module("hyperparameter_tuning")
evaluate_metrics = import_custom_module("evaluate_metrics")

# Assign functions
train_and_evaluate = model_training.train_and_evaluate
run_grid_search = hyperparameter_tuning.run_grid_search
run_random_search = hyperparameter_tuning.run_random_search
plot_confusion_matrix = evaluate_metrics.plot_confusion_matrix
generate_classification_report = evaluate_metrics.generate_classification_report
```
